```

public interface IPlayable { void Play(); }
public interface IAwardable { void WinAward(string award); }

public class Movie : IPlayable, IAwardable
{
    public string Title { get; set; }
    public string Director { get; set; }
    public int ReleaseYear { get; set; }
    public int DurationMinutes { get; set; }
    public double Rating { get; set; }

    public Movie(string title, string director, int year, int duration, double rating)
    {
        Title = title;
        Director = director;
        ReleaseYear = year;
        DurationMinutes = duration;
        Rating = rating;
    }

    public virtual string GetInfo() => $"{Title}, {Director}, {ReleaseYear}г., {DurationMinutes} мин., Рейтинг: {Rating}";

    public virtual void Watch() => Console.WriteLine($"Смотрим фильм: {Title}");
    public virtual void Rate() => Console.WriteLine($"Оценка фильма {Title}: {Rating}");

    public void Watch(int startMinute) => Console.WriteLine($"Смотрим фильм {Title} с {startMinute}-й минуты.");

    public void Rate(double userRating)
    {
        Rating = (Rating + userRating) / 2;
        Console.WriteLine($"Новый средний рейтинг фильма {Title}: {Rating:F1}");
    }

    public void Play() => Console.WriteLine($"Старт воспроизведения: {Title}");
    public void WinAward(string award) => Console.WriteLine($"Фильм {Title} получил награду: {award}");
}

public class Documentary : Movie
{
    public string Theme { get; set; }
    public string Narrator { get; set; }

    public Documentary(string title, string director, int year, int duration, double rating, string theme, string narrator)
        : base(title, director, year, duration, rating)
    {
        Theme = theme;
        Narrator = narrator;
    }

    public override void Watch() => Console.WriteLine($"Документальный фильм: {Title}, Тема: {Theme}, Диктор: {Narrator}");
}

public class FeatureFilm : Movie
{
    public string Genre { get; set; }
    public decimal Budget { get; set; }

    public FeatureFilm(string title, string director, int year, int duration, double rating, string genre, decimal budget)
        : base(title, director, year, duration, rating)
    {
        Genre = genre;
        Budget = budget;
    }

    public override void Rate() => Console.WriteLine($"Художественный фильм ({Genre}) {Title} имеет оценку {Rating}/10 (Бюджет: ${Budget:N0})");
}

public class AnimatedMovie : Movie
{
    public string AnimationStudio { get; set; }
    public bool Is3D { get; set; }

    public AnimatedMovie(string title, string director, int year, int duration, double rating, string studio, bool is3D)
        : base(title, director, year, duration, rating)
    {
        AnimationStudio = studio;
        Is3D = is3D;
    }

    public override string GetInfo() => base.GetInfo() + $", Студия: {AnimationStudio}, 3D: {(Is3D ? "Да" : "Нет")}";
}

public class AnimeMovie : AnimatedMovie
{
    public string OriginalManga { get; set; }
    public string DubStudio { get; set; }

    public AnimeMovie(string title, string director, int year, int duration, double rating, string studio, bool is3D, string manga, string dubStudio)
        : base(title, director, year, duration, rating, studio, is3D)
    {
        OriginalManga = manga;
        DubStudio = dubStudio;
    }

    public override string GetInfo() => base.GetInfo() + $", Манга: {OriginalManga}, Озвучка: {DubStudio}";
}

public class MovieCollection<T> where T : Movie
{
    private List<T> _movies = new List<T>();

    public void Add(T movie) => _movies.Add(movie);
    public void Remove(T movie) => _movies.Remove(movie);

    public void DisplayCatalog()
    {
        foreach (var m in _movies)
        {
            Console.WriteLine(m.GetInfo());
            m.Watch();
            m.Rate();
            Console.WriteLine();
        }
    }
}
```

```
MovieCollection<Movie> library = new MovieCollection<Movie>();

library.Add(new Documentary("Земля", "Алексей", 2007, 90, 8.5, "Природа", "Николай Дроздов"));
library.Add(new FeatureFilm("Матрица", "Вачовски", 1999, 136, 8.7, "Фантастика", 63000000m));
library.Add(new AnimatedMovie("Король Лев", "Минкофф", 1994, 88, 8.8, "Disney", false));
library.Add(new AnimeMovie("Унесенные призраками", "Миядзаки", 2001, 125, 8.6, "Ghibli", false, "Оригинал", "Reamedia"));

library.DisplayCatalog();

Console.WriteLine("Демонстрация перегрузок методов");
Movie sampleMovie = new FeatureFilm("Интерстеллар", "Кристофер Нолан", 2014, 169, 8.6, "Фантастика", 165000000m);
sampleMovie.Watch(45);
sampleMovie.Rate(10.0);
```
