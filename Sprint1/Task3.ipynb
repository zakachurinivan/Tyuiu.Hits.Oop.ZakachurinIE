Создать базовый класс Movie в C#, который будет представлять информацию о
фильмах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;

public class Movie
{
    private string _title;
    private string _director;
    private int _releaseYear;

    public string Title
    {
        get { return _title; }
        set { _title = value; }
    }

    public string Director
    {
        get { return _director; }
        set { _director = value; }
    }

    public int ReleaseYear
    {
        get { return _releaseYear; }
        set
        {
            if (value >= 1888 && value <= DateTime.Now.Year + 5)
                _releaseYear = value;
            else
                throw new ArgumentOutOfRangeException("Некорректный год выпуска!");
        }
    }

    public Movie()
    {
        _title = "Без названия";
        _director = "Неизвестен";
        _releaseYear = 2000;
    }

    public Movie(string title, string director, int year) : this()
    {
        Title = title;
        Director = director;
        ReleaseYear = year;
    }

    public virtual string GetInfo()
    {
        return $"{Title}, {Director}, {ReleaseYear}";
    }

    public virtual void Watch()
    {
        Console.WriteLine("Смотрим фильм: " + Title);
    }

    public virtual void Rate()
    {
        Console.WriteLine("Оценка фильма: " + Title);
    }

    public void CompareWith(Movie other)
    {
        if (other == null)
        {
            Console.WriteLine("Сравнение невозможно: фильм не указан.");
            return;
        }

        Console.WriteLine($"Сравнение: «{Title}» ({ReleaseYear}) и «{other.Title}» ({other.ReleaseYear})");

        if (ReleaseYear > other.ReleaseYear)
            Console.WriteLine($"«{Title}» выпущен позже.");
        else if (ReleaseYear < other.ReleaseYear)
            Console.WriteLine($"«{other.Title}» выпущен позже.");
        else
            Console.WriteLine("Фильмы выпущены в один год.");
    }
}

public class Documentary : Movie
{
    private string _theme;

    public string Theme
    {
        get { return _theme; }
        set { _theme = value; }
    }

    public Documentary() : base()
    {
        _theme = "Общая";
    }

    public Documentary(string title, string director, int year, string theme)
        : base(title, director, year)
    {
        Theme = theme;
    }

    public override void Watch()
    {
        Console.WriteLine($"Документальный фильм: {Title}, Тема: {Theme}");
    }

    public void Explore()
    {
        Console.WriteLine($"Изучаем тему «{Theme}» в фильме «{Title}».");
    }
}

public class FeatureFilm : Movie
{
    private string _genre;

    public string Genre
    {
        get { return _genre; }
        set { _genre = value; }
    }

    public FeatureFilm() : base()
    {
        _genre = "Драма";
    }

    public FeatureFilm(string title, string director, int year, string genre)
        : base(title, director, year)
    {
        Genre = genre;
    }

    public override void Rate()
    {
        Console.WriteLine($"Оценка жанра {Genre}: высокая");
    }

    public void RecommendTo(Movie other)
    {
        Console.WriteLine($"Рекомендую фильм «{Title}» (жанр: {Genre}) к просмотру вместе с «{other.Title}».");
    }
}

public class AnimatedMovie : Movie
{
    private string _animationStudio;

    public string AnimationStudio
    {
        get { return _animationStudio; }
        set { _animationStudio = value; }
    }

    public AnimatedMovie() : base()
    {
        _animationStudio = "Неизвестная студия";
    }

    public AnimatedMovie(string title, string director, int year, string studio)
        : base(title, director, year)
    {
        AnimationStudio = studio;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Студия: {AnimationStudio}";
    }

    public void CollaborateWith(AnimatedMovie other)
    {
        Console.WriteLine($"Студия «{AnimationStudio}» совместно со студией «{other.AnimationStudio}» " +
                          $"работают над проектами «{Title}» и «{other.Title}».");
    }
}

class Program
{
    static void Main()
    {
        Movie[] movies = new Movie[]
        {
            new Documentary("Земля", "Алексей", 2007, "Природа"),
            new FeatureFilm("Матрица", "Вачовски", 1999, "Фантастика"),
            new AnimatedMovie("Король Лев", "Минкофф", 1994, "Disney")
        };

        foreach (var m in movies)
        {
            Console.WriteLine(m.GetInfo());
            m.Watch();
            m.Rate();
            Console.WriteLine();
        }

        Console.WriteLine("=== Взаимодействие объектов ===\n");

        movies[0].CompareWith(movies[1]);

        if (movies[1] is FeatureFilm feature)
            feature.RecommendTo(movies[2]);

        if (movies[2] is AnimatedMovie animated)
        {
            var another = new AnimatedMovie("Шрек", "Адамсон", 2001, "DreamWorks");
            animated.CollaborateWith(another);
        }

        if (movies[0] is Documentary doc)
            doc.Explore();
    }
}
```
