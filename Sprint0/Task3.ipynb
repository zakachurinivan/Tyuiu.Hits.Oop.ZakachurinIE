<h2 style="color:DodgerBlue">Описание проекта:</h2>

----
Создать базовый класс Movie в C#, который будет представлять информацию о
фильмах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.



<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public class Movie
{
    public string Title;
    public string Director;
    public int ReleaseYear;

    public Movie(string title, string director, int year)
    {
        Title = title;
        Director = director;
        ReleaseYear = year;
    }

    public virtual string GetInfo()
    {
        return $"{Title}, {Director}, {ReleaseYear}";
    }

    public virtual void Watch()
    {
        Console.WriteLine("Смотрим фильм: " + Title);
    }

    public virtual void Rate()
    {
        Console.WriteLine("Оценка фильма: " + Title);
    }
}

public class Documentary : Movie
{
    public string Theme;

    public Documentary(string title, string director, int year, string theme)
        : base(title, director, year)
    {
        Theme = theme;
    }

    public override void Watch()
    {
        Console.WriteLine($"Документальный фильм: {Title}, Тема: {Theme}");
    }
}

public class FeatureFilm : Movie
{
    public string Genre;

    public FeatureFilm(string title, string director, int year, string genre)
        : base(title, director, year)
    {
        Genre = genre;
    }

    public override void Rate()
    {
        Console.WriteLine($"Оценка жанра {Genre}: высокая");
    }
}

public class AnimatedMovie : Movie
{
    public string AnimationStudio;

    public AnimatedMovie(string title, string director, int year, string studio)
        : base(title, director, year)
    {
        AnimationStudio = studio;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Студия: {AnimationStudio}";
    }
}
```

```
Movie[] movies = new Movie[]
{
    new Documentary("Земля", "Алексей", 2007, "Природа"),
    new FeatureFilm("Матрица", "Вачовски", 1999, "Фантастика"),
    new AnimatedMovie("Король Лев", "Минкофф", 1994, "Disney")
};
foreach (var m in movies)
{
    Console.WriteLine(m.GetInfo());
    m.Watch();
    m.Rate();
    Console.WriteLine();
}
```
