```
using System;

public interface IPlayable { void Play(); }
public interface IAwardable { void WinAward(string award); }

public class Movie : IPlayable, IAwardable
{
    public string Title { get; set; }
    public string Director { get; set; }
    public int ReleaseYear { get; set; }

    public Movie(string title, string director, int year)
    {
        Title = title;
        Director = director;
        ReleaseYear = year;
    }

    public virtual string GetInfo() => $"{Title}, {Director}, {ReleaseYear}";
    public virtual void Watch() => Console.WriteLine($"Смотрим фильм: {Title}");
    public virtual void Rate() => Console.WriteLine($"Оценка фильма: {Title}");
    public void Play() => Console.WriteLine($"Старт воспроизведения: {Title}");
    public void WinAward(string award) => Console.WriteLine($"Фильм {Title} получил награду: {award}");
}

public class Documentary : Movie
{
    public string Theme { get; set; }
    public Documentary(string title, string director, int year, string theme) 
        : base(title, director, year) => Theme = theme;

    public override void Watch() => Console.WriteLine($"Документальный фильм: {Title}, Тема: {Theme}");
}

public class FeatureFilm : Movie
{
    public string Genre { get; set; }
    public FeatureFilm(string title, string director, int year, string genre) 
        : base(title, director, year) => Genre = genre;

    public override void Rate() => Console.WriteLine($"Оценка жанра {Genre}: высокая");
}

public class AnimatedMovie : Movie
{
    public string AnimationStudio { get; set; }
    public AnimatedMovie(string title, string director, int year, string studio) 
        : base(title, director, year) => AnimationStudio = studio;

    public override string GetInfo() => base.GetInfo() + $", Студия: {AnimationStudio}";
}

public class AnimeMovie : AnimatedMovie
{
    public string OriginalManga { get; set; }
    public AnimeMovie(string title, string director, int year, string studio, string manga) 
        : base(title, director, year, studio) => OriginalManga = manga;

    public override string GetInfo() => base.GetInfo() + $", Манга: {OriginalManga}";
}
```

```
Movie[] movies = new Movie[]
{
    new Documentary("Земля", "Алексей", 2007, "Природа"),
    new FeatureFilm("Матрица", "Вачовски", 1999, "Фантастика"),
    new AnimatedMovie("Король Лев", "Минкофф", 1994, "Disney"),
    new AnimeMovie("Унесенные призраками", "Миядзаки", 2001, "Ghibli", "Оригинал")
};

foreach (var m in movies)
{
    Console.WriteLine(m.GetInfo());
    m.Watch();
    m.Rate();
    Console.WriteLine();
}
```
